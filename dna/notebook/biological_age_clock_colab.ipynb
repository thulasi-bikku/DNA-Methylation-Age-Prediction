```python
import pandas as pd
import numpy as np

# Load our clean age lookup table
age_df = pd.read_csv("age_metadata.csv")
age_lookup = dict(zip(age_df["Sample_ID"], age_df["Age"]))

chunk_size = 5000  # process 5000 CpG rows at a time
correlations = []

reader = pd.read_csv("GSE40279_average_beta.txt", sep="\t", chunksize=chunk_size)

chunk_num = 0
for chunk in reader:
    chunk_num += 1
    chunk = chunk.set_index("ID_REF")

    # Align column order to our age list, in case of any mismatch
    sample_cols = [c for c in chunk.columns if c in age_lookup]
    ages_array = np.array([age_lookup[c] for c in sample_cols])

    # Compute correlation of each CpG row with age
    for cpg_id, row in chunk[sample_cols].iterrows():
        values = row.values.astype(float)
        if np.isnan(values).any():
            continue
        corr = np.corrcoef(values, ages_array)[0, 1]
        correlations.append((cpg_id, corr))

    if chunk_num % 20 == 0:
        print(f"Processed {chunk_num * chunk_size} CpG sites so far...")

print("Done. Total CpGs processed:", len(correlations))
```

```python
import pandas as pd

# Convert our correlation results into a dataframe
corr_df = pd.DataFrame(correlations, columns=["CpG_ID", "Correlation"])

# We care about strength of relationship in EITHER direction (positive or negative),
# since both hyper- and hypo-methylation with age are biologically meaningful
corr_df["Abs_Correlation"] = corr_df["Correlation"].abs()

# Sort by strongest relationship first
corr_df = corr_df.sort_values("Abs_Correlation", ascending=False).reset_index(drop=True)

# Save the full ranked list permanently (useful later for the paper's supplementary table)
corr_df.to_csv("cpg_age_correlation_ranked.csv", index=False)

print("Top 20 most age-correlated CpG sites:")
print(corr_df.head(20))

# Select top 3000 candidate CpGs for model training
top_n = 3000
top_cpgs = corr_df.head(top_n)["CpG_ID"].tolist()
print(f"\nSelected top {top_n} CpGs for training.")
```

```python
import pandas as pd

# Re-read the beta file, but only keeping rows matching our selected top CpGs
chunk_size = 5000
selected_rows = []

reader = pd.read_csv("GSE40279_average_beta.txt", sep="\t", chunksize=chunk_size)

top_cpgs_set = set(top_cpgs)

for chunk in reader:
    chunk = chunk.set_index("ID_REF")
    matched = chunk[chunk.index.isin(top_cpgs_set)]
    if not matched.empty:
        selected_rows.append(matched)

# Combine all matched chunks into one dataframe: CpGs as rows, patients as columns
beta_subset = pd.concat(selected_rows)
print("Shape (CpGs x Patients):", beta_subset.shape)

# Transpose so patients become rows, CpGs become columns
final_df = beta_subset.T
final_df.index.name = "Sample_ID"
final_df = final_df.reset_index()

# Merge in the Age column using our earlier lookup table
final_df = final_df.merge(age_df, on="Sample_ID", how="inner")

print("Final training table shape (Patients x [CpGs + Age]):", final_df.shape)
final_df.head()
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error
import time

# Load top 300 candidates from our earlier correlation ranking
corr_ranked = pd.read_csv("cpg_age_correlation_ranked.csv")
candidate_cpgs = corr_ranked.head(300)["CpG_ID"].tolist()

importance_scores = []
start_time = time.time()

for i, cpg in enumerate(candidate_cpgs):
    # Make a copy of the data, shuffle just this one column
    shuffled_df = predict_df.copy()
    shuffled_df[cpg] = np.random.permutation(shuffled_df[cpg].values)

    # Predict with the shuffled column
    shuffled_result = model.predict(shuffled_df)
    shuffled_preds = shuffled_result[0]["Age_predictions"].values
    shuffled_mae = mean_absolute_error(shuffled_df["Age"], shuffled_preds)

    # Importance = how much worse the error got
    importance = shuffled_mae - baseline_mae
    importance_scores.append((cpg, importance))

    if (i+1) % 25 == 0:
        elapsed = time.time() - start_time
        print(f"Processed {i+1}/300 CpGs... ({elapsed:.1f}s elapsed)")

print("Permutation importance calculation complete.")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

age_df = pd.read_csv("age_metadata.csv")
print("Total samples:", len(age_df))

# Split OFF the final test set FIRST (20%) — this is never touched again until final evaluation
trainval_ids, test_ids = train_test_split(
    age_df["Sample_ID"].values,
    test_size=0.20,
    random_state=42
)

print(f"Train+Val pool: {len(trainval_ids)} samples")
print(f"Held-out Test: {len(test_ids)} samples")

# Save these ID lists permanently so every later step uses the EXACT same split
pd.Series(trainval_ids).to_csv("trainval_ids.csv", index=False)
pd.Series(test_ids).to_csv("test_ids.csv", index=False)
```

```python
trainval_set = set(trainval_ids)
age_lookup = dict(zip(age_df["Sample_ID"], age_df["Age"]))

chunk_size = 5000
correlations = []

reader = pd.read_csv("GSE40279_average_beta.txt", sep="\t", chunksize=chunk_size)

chunk_num = 0
for chunk in reader:
    chunk_num += 1
    chunk = chunk.set_index("ID_REF")

    # CRITICAL FIX: only use columns belonging to the train+val pool
    sample_cols = [c for c in chunk.columns if c in trainval_set]
    ages_array = np.array([age_lookup[c] for c in sample_cols])

    for cpg_id, row in chunk[sample_cols].iterrows():
        values = row.values.astype(float)
        if np.isnan(values).any():
            continue
        corr = np.corrcoef(values, ages_array)[0, 1]
        correlations.append((cpg_id, corr))

    if chunk_num % 20 == 0:
        print(f"Processed {chunk_num * chunk_size} CpG sites so far...")

print("Done. Total CpGs processed:", len(correlations))

corr_df = pd.DataFrame(correlations, columns=["CpG_ID", "Correlation"])
corr_df["Abs_Correlation"] = corr_df["Correlation"].abs()
corr_df = corr_df.sort_values("Abs_Correlation", ascending=False).reset_index(drop=True)
corr_df.to_csv("cpg_correlation_TRAINONLY.csv", index=False)

top_cpgs = corr_df.head(3000)["CpG_ID"].tolist()
print("Top CpG (train-only):", corr_df.iloc[0]["CpG_ID"], corr_df.iloc[0]["Correlation"])
```

```python
chunk_size = 5000
selected_rows = []
top_cpgs_set = set(top_cpgs)

reader = pd.read_csv("GSE40279_average_beta.txt", sep="\t", chunksize=chunk_size)
for chunk in reader:
    chunk = chunk.set_index("ID_REF")
    matched = chunk[chunk.index.isin(top_cpgs_set)]
    if not matched.empty:
        selected_rows.append(matched)

beta_subset = pd.concat(selected_rows)
full_df = beta_subset.T
full_df.index.name = "Sample_ID"
full_df = full_df.reset_index()
full_df = full_df.merge(age_df, on="Sample_ID", how="inner")

# Split into the two pools using our fixed ID lists
trainval_df = full_df[full_df["Sample_ID"].isin(trainval_ids)].reset_index(drop=True)
test_df = full_df[full_df["Sample_ID"].isin(test_ids)].reset_index(drop=True)

trainval_df.to_csv("trainval_data_leakfree.csv", index=False)
test_df.to_csv("test_data_leakfree.csv", index=False)

print("Train+Val shape:", trainval_df.shape)
print("Test shape:", test_df.shape)
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import ElasticNetCV
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error
import xgboost as xgb
import lightgbm as lgb

# Load leak-free data
trainval_df = pd.read_csv("trainval_data_leakfree.csv")
test_df = pd.read_csv("test_data_leakfree.csv")

feature_cols = [c for c in trainval_df.columns if c.startswith("cg")]
X_train = trainval_df[feature_cols].values
y_train = trainval_df["Age"].values
X_test = test_df[feature_cols].values
y_test = test_df["Age"].values

print("Train shape:", X_train.shape, "Test shape:", X_test.shape)

results = {}

def evaluate(name, preds):
    mae = mean_absolute_error(y_test, preds)
    r2 = r2_score(y_test, preds)
    rmse = np.sqrt(mean_squared_error(y_test, preds))
    results[name] = {"MAE": mae, "R2": r2, "RMSE": rmse}
    print(f"{name}: MAE={mae:.3f}, R2={r2:.3f}, RMSE={rmse:.3f}")

# ---- Baseline 1: Elastic Net (mandatory) ----
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

enet = ElasticNetCV(cv=5, random_state=42, n_jobs=-1, max_iter=5000)
enet.fit(X_train_scaled, y_train)
evaluate("Elastic Net", enet.predict(X_test_scaled))

# ---- Baseline 2: XGBoost ----
xgb_model = xgb.XGBRegressor(n_estimators=300, max_depth=4, learning_rate=0.05, random_state=42, n_jobs=-1)
xgb_model.fit(X_train, y_train)
evaluate("XGBoost", xgb_model.predict(X_test))

# ---- Baseline 3: LightGBM ----
lgb_model = lgb.LGBMRegressor(n_estimators=300, max_depth=4, learning_rate=0.05, random_state=42, n_jobs=-1, verbose=-1)
lgb_model.fit(X_train, y_train)
evaluate("LightGBM", lgb_model.predict(X_test))

# ---- Baseline 4: Simple MLP ----
mlp = MLPRegressor(hidden_layer_sizes=(128, 64), max_iter=1000, random_state=42, early_stopping=True)
mlp.fit(X_train_scaled, y_train)
evaluate("Simple MLP", mlp.predict(X_test_scaled))

print("\nAll baseline results:")
for name, metrics in results.items():
    print(name, metrics)

# Save results permanently
pd.DataFrame(results).T.to_csv("baseline_model_results.csv")
```

```python
from sklearn.neural_network import MLPRegressor

best_mlp_mae = None
best_alpha = None

for alpha_val in [0.001, 0.01, 0.1, 1.0, 5.0]:
    mlp_test = MLPRegressor(
        hidden_layer_sizes=(128, 64),
        alpha=alpha_val,
        max_iter=2000,
        random_state=42,
        early_stopping=True,
        validation_fraction=0.15
    )
    mlp_test.fit(X_train_scaled, y_train)
    preds = mlp_test.predict(X_test_scaled)
    mae = mean_absolute_error(y_test, preds)
    r2 = r2_score(y_test, preds)
    print(f"alpha={alpha_val}: MAE={mae:.3f}, R2={r2:.3f}")
    if best_mlp_mae is None or mae < best_mlp_mae:
        best_mlp_mae = mae
        best_alpha = alpha_val

print(f"\nBest alpha: {best_alpha}, Best MAE: {best_mlp_mae:.3f}")

# Update results with the properly regularized MLP
final_mlp = MLPRegressor(hidden_layer_sizes=(128, 64), alpha=best_alpha, max_iter=2000, random_state=42, early_stopping=True, validation_fraction=0.15)
final_mlp.fit(X_train_scaled, y_train)
evaluate("Simple MLP", final_mlp.predict(X_test_scaled))

pd.DataFrame(results).T.to_csv("baseline_model_results.csv")
print(pd.DataFrame(results).T)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.linear_model import ElasticNetCV
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, r2_score
import xgboost as xgb
import lightgbm as lgb

trainval_df = pd.read_csv("trainval_data_leakfree.csv")
feature_cols = [c for c in trainval_df.columns if c.startswith("cg")]
X = trainval_df[feature_cols].values
y = trainval_df["Age"].values

kf = KFold(n_splits=5, shuffle=True, random_state=42)

cv_results = {"Elastic Net": [], "XGBoost": [], "LightGBM": [], "Simple MLP": []}

fold_num = 0
for train_idx, val_idx in kf.split(X):
    fold_num += 1
    X_tr, X_val = X[train_idx], X[val_idx]
    y_tr, y_val = y[train_idx], y[val_idx]

    scaler = StandardScaler()
    X_tr_scaled = scaler.fit_transform(X_tr)
    X_val_scaled = scaler.transform(X_val)

    enet = ElasticNetCV(cv=3, random_state=42, n_jobs=-1, max_iter=5000)
    enet.fit(X_tr_scaled, y_tr)
    cv_results["Elastic Net"].append(mean_absolute_error(y_val, enet.predict(X_val_scaled)))

    xgb_model = xgb.XGBRegressor(n_estimators=300, max_depth=4, learning_rate=0.05, random_state=42, n_jobs=-1)
    xgb_model.fit(X_tr, y_tr)
    cv_results["XGBoost"].append(mean_absolute_error(y_val, xgb_model.predict(X_val)))

    lgb_model = lgb.LGBMRegressor(n_estimators=300, max_depth=4, learning_rate=0.05, random_state=42, n_jobs=-1, verbose=-1)
    lgb_model.fit(X_tr, y_tr)
    cv_results["LightGBM"].append(mean_absolute_error(y_val, lgb_model.predict(X_val)))

    mlp = MLPRegressor(hidden_layer_sizes=(128,64), alpha=0.1, max_iter=2000, random_state=42, early_stopping=True, validation_fraction=0.15)
    mlp.fit(X_tr_scaled, y_tr)
    cv_results["Simple MLP"].append(mean_absolute_error(y_val, mlp.predict(X_val_scaled)))

    print(f"Fold {fold_num} done.")

print("\n5-Fold CV Results (Mean ± Std MAE):")
cv_summary = {}
for model, maes in cv_results.items():
    mean_mae = np.mean(maes)
    std_mae = np.std(maes)
    cv_summary[model] = {"Mean_MAE": mean_mae, "Std_MAE": std_mae, "Fold_MAEs": maes}
    print(f"{model}: {mean_mae:.3f} ± {std_mae:.3f}  (folds: {[round(m,2) for m in maes]})")

pd.DataFrame(cv_summary).T.to_csv("cv_results_baselines.csv")
```

```python
import pandas as pd
import numpy as np
from scipy import stats

# Reload baseline models trained on full trainval, predict on the same held-out test set
trainval_df = pd.read_csv("trainval_data_leakfree.csv")
test_df = pd.read_csv("test_data_leakfree.csv")
feature_cols = [c for c in trainval_df.columns if c.startswith("cg")]

X_train = trainval_df[feature_cols].values
y_train = trainval_df["Age"].values
X_test = test_df[feature_cols].values
y_test = test_df["Age"].values

from sklearn.linear_model import ElasticNetCV
from sklearn.preprocessing import StandardScaler
import xgboost as xgb
import lightgbm as lgb

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

enet = ElasticNetCV(cv=5, random_state=42, n_jobs=-1, max_iter=5000)
enet.fit(X_train_scaled, y_train)
enet_preds = enet.predict(X_test_scaled)

xgb_model = xgb.XGBRegressor(n_estimators=300, max_depth=4, learning_rate=0.05, random_state=42, n_jobs=-1)
xgb_model.fit(X_train, y_train)
xgb_preds = xgb_model.predict(X_test)

lgb_model = lgb.LGBMRegressor(n_estimators=300, max_depth=4, learning_rate=0.05, random_state=42, n_jobs=-1, verbose=-1)
lgb_model.fit(X_train, y_train)
lgb_preds = lgb_model.predict(X_test)

# FT-Transformer predictions (already have these as 'preds' from before)
ft_preds = preds

# Per-sample absolute errors for each model
ft_errors = np.abs(y_test - ft_preds)
enet_errors = np.abs(y_test - enet_preds)
xgb_errors = np.abs(y_test - xgb_preds)
lgb_errors = np.abs(y_test - lgb_preds)

print("Mean absolute errors (sanity check):")
print(f"FT-Transformer: {ft_errors.mean():.3f}")
print(f"Elastic Net: {enet_errors.mean():.3f}")
print(f"XGBoost: {xgb_errors.mean():.3f}")
print(f"LightGBM: {lgb_errors.mean():.3f}")
```

```python
print("\n--- Paired Statistical Tests: FT-Transformer vs. Baselines ---\n")

for name, baseline_errors in [("Elastic Net", enet_errors), ("XGBoost", xgb_errors), ("LightGBM", lgb_errors)]:
    t_stat, t_pval = stats.ttest_rel(ft_errors, baseline_errors)
    w_stat, w_pval = stats.wilcoxon(ft_errors, baseline_errors)

    print(f"FT-Transformer vs {name}:")
    print(f"  Paired t-test:      t={t_stat:.3f}, p={t_pval:.6f}")
    print(f"  Wilcoxon test:      W={w_stat:.3f}, p={w_pval:.6f}")
    significant = "SIGNIFICANT (p<0.05)" if t_pval < 0.05 else "not significant"
    print(f"  Result: {significant}\n")

# Save all results
sig_results = pd.DataFrame({
    "FT-Transformer": ft_errors,
    "Elastic Net": enet_errors,
    "XGBoost": xgb_errors,
    "LightGBM": lgb_errors
})
sig_results.to_csv("per_sample_errors_all_models.csv", index=False)
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import ElasticNetCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error
import xgboost as xgb

# Load the train-only correlation ranking (already leak-free)
corr_df = pd.read_csv("cpg_correlation_TRAINONLY.csv")
trainval_df = pd.read_csv("trainval_data_leakfree.csv")
test_df = pd.read_csv("test_data_leakfree.csv")

feature_counts = [500, 1000, 3000]
ablation_results = {}

for n_feat in feature_counts:
    top_n_cpgs = corr_df.head(n_feat)["CpG_ID"].tolist()
    available_cpgs = [c for c in top_n_cpgs if c in trainval_df.columns]

    X_train = trainval_df[available_cpgs].values
    y_train = trainval_df["Age"].values
    X_test = test_df[available_cpgs].values
    y_test = test_df["Age"].values

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # Elastic Net at this feature count
    enet = ElasticNetCV(cv=5, random_state=42, n_jobs=-1, max_iter=5000)
    enet.fit(X_train_scaled, y_train)
    enet_mae = mean_absolute_error(y_test, enet.predict(X_test_scaled))

    # XGBoost at this feature count
    xgb_model = xgb.XGBRegressor(n_estimators=300, max_depth=4, learning_rate=0.05, random_state=42, n_jobs=-1)
    xgb_model.fit(X_train, y_train)
    xgb_mae = mean_absolute_error(y_test, xgb_model.predict(X_test))

    ablation_results[n_feat] = {"Elastic_Net_MAE": enet_mae, "XGBoost_MAE": xgb_mae, "Features_Used": len(available_cpgs)}
    print(f"n={n_feat}: Elastic Net MAE={enet_mae:.3f}, XGBoost MAE={xgb_mae:.3f}")

ablation_df = pd.DataFrame(ablation_results).T
ablation_df.to_csv("feature_ablation_results.csv")
print("\nFull ablation results:")
print(ablation_df)
```
